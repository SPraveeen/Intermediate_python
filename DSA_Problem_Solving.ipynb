## count the factors of number N

```python
# normal method
# by default system do 1 sec for 10^8
# if it's going to be 10^9 => 10^8 * 10 => 10 sec
# but if 10 ^ 18 => 10^8 * 10^10 => almost 317 years
n=int(input('Enter a number'))
count=0
for i in range(1,n+1):
    if n%i==0:
        count+=1
print(count)
```

```python
# optimization approach is
# N/i=?
# N=?*i
# for 25 => 1,5,25 chk in chatgpt
from math import floor
n=int(input('Enter a number'))
count=0
for i in range(1,floor(sqrt(n))+1):
    if n%i == 0:
        if i!=n//i:
            count+=2
        else:
            count+=1
print(count)
```

## prime numbers

```python
from math import floor
n=int(input('Enter a number'))
count=0
for i in range(1,floor(sqrt(n))+1):
    if n%i == 0:
        if i!=n//i:
            count+=2
        else:
            count+=1
print(count)

if count==2:
    print('prime_number')
else:
    print('Not a prime number')
```
