```python
def triangle_find(num1,num2,num3):
    if num1+num2 > num3 and num2+num3 > num1 and num3+num1 > num2:
        return "forms triangle"
    else:
        return "not"
```

```python
# WAP to generate the next 15 leap years starting from a given year. populate the leap years into a list and display the list
```

```python
# leap year
number=int(input('Enter year'))
lis=[]
while len(lis) < 15:
    if (number%400==0) or (number%4==0 and number%100 != 0):
        lis.append(number)
    number+=1
print(lis)
```

```python
def is_leap_year(number):
    if (number%400==0) or (number%4==0 and number%100 != 0):
        return True
    return False

def next_15(start_year):
    leap_year=[]
    year=start_year
    while len(leap_year) < 15:
        if is_leap_year(year):
            leap_year.append(year)
        year+=1
```
